```
# Basic List Operation
fruits = ["apple","banana","orange","mango"]
print("Original List : ",fruits)
print("First Fruit : ",fruits[0])
print("Last Fruit : ",fruits[-1])
```

```
# Adding Items
fruits.append("kiwi")
print("Updated list : ",fruits)
```

```
# Removing Items
fruits.remove("apple")
print("Updated list : ",fruits)
```

```
#Slicing
print("Frist 3 fruits")
fruits[:3]
```

```
#List with numbers
numbers=[5,2,1,8,9,3]
print("Sum: ",sum(numbers))
print("Max: ",max(numbers))
print("Min: ",min(numbers))
print("Sorted: ",sorted(numbers))
print("Length: ",len(numbers))
print("Reverse: ",sorted(numbers,reverse=True))
```

```
# Basic dictionary Operations

student = {
  "name": "Avash",
  "age": 21,
  "course": "computer",
  "grade": "A+"
}
print("Student : ",student)
print("Student name : ",student["name"])
print("Keys : ",student.keys())
print("Values : ",student.values())
```

```
# Adding or Updating
student["grade"] = "A"
student["city"] = "Kathmandu"
print("Updated student : ",student)
```

```
# Task 1 : create a dictionary of 3 product and their price
products = {
  "product1": 100,
  "product2": 200,
  "product3": 300
}
print("Products : ",products)
```

```
# Add a new product
products["product4"] = 400
print("Updated products : ",products)
```

```
# Update product
products["product2"] = 500
print("Updated products : ",products)
```

```
# Display all products
for product, price in products.items():
  print(f"{product}:${price}")
```

```
# Tuples (inmutable,ordered)
# Create a tuple of days in a week
days = ("Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday")
# using positive logic
print("weekend : ",days[5:7])
print("weekdays : ",days[:5])
print("First day : ",days[0])
# using negative logic
print("Last day : ",days[-1])
print("weekend : ",days[-2:])
print("weekdays : ",days[:-2])
```
